```python
import pandas as pd
import re
from pathlib import Path

# Read the CSV file
df = pd.read_csv(Path(__file__).resolve().parents[1] / 'data' / 'Detailed Table.csv', header=2)

# Function to extract the last number from a string
def extract_last_number(s):
    if isinstance(s, str):
        numbers = re.findall(r'\d+', s)
        if numbers:
            return int(numbers[-1])
    return s

# Apply the function to all cells in the dataframe
df = df.applymap(extract_last_number)

# Save the cleaned data to a new CSV file
df.to_csv(Path(__file__).resolve().parents[1] / 'data' / 'cleaned_detailed_table.csv', index=False)

print('Cleaned data saved to cleaned_detailed_table.csv')
```
