```python
# Generate gamma parameters - method of moments
import pandas as pd
import numpy as np
from scipy.stats import gamma

# Load the CSV
data = pd.read_csv("/mnt/data/Microcosting20230814PSA.csv")

# Filter out rows with "-" and convert to numeric
filtered_data = data[(data["LL"] != "-") & (data["MP"] != "-") & (data["UL"] != "-")]
filtered_data["LL"] = pd.to_numeric(filtered_data["LL"])
filtered_data["MP"] = pd.to_numeric(filtered_data["MP"])
filtered_data["UL"] = pd.to_numeric(filtered_data["UL"])

# Function to estimate gamma parameters using MLE
def estimate_gamma_parameters(row):
    shape, loc, scale = gamma.fit([row["LL"], row["UL"]], floc=0)
    return pd.Series([shape, scale])

# Add "Shape" and "Scale" columns and estimate parameters
filtered_data["Shape"], filtered_data["Scale"] = [float('nan')] * 2
filtered_data = filtered_data.dropna(subset=["LL", "UL"])
filtered_data[["Shape", "Scale"]] = filtered_data.apply(estimate_gamma_parameters, axis=1)

# Export the updated table to a new CSV file
output_path = "/mnt/data/Updated_Microcosting_MLE.csv"
filtered_data[["Test", "Category", "Stage", "LL", "MP", "UL", "Shape", "Scale"]].to_csv(output_path, index=False)
```

```python
# Generate gamma parameters - maximum likelihood estimation method
import pandas as pd
import numpy as np
from scipy.stats import gamma
from scipy.optimize import minimize

# Load the CSV
data = pd.read_csv("/mnt/data/Microcosting20230814PSA.csv")

# Filter out rows with "-" and convert to numeric
filtered_data = data[(data["LL"] != "-") & (data["MP"] != "-") & (data["UL"] != "-")]
filtered_data["LL"] = pd.to_numeric(filtered_data["LL"])
filtered_data["MP"] = pd.to_numeric(filtered_data["MP"])
filtered_data["UL"] = pd.to_numeric(filtered_data["UL"])

def negative_log_likelihood(params, data):
    """ Negative log-likelihood for the gamma distribution. """
    shape, scale = params
    return -np.sum(gamma.logpdf(data, shape, scale=scale))

def estimate_gamma_mle(data):
    """ Estimate gamma distribution parameters using MLE. """
    # Initial guess for parameters (method of moments)
    mean_data = np.mean(data)
    std_data = np.std(data)
    shape_guess = mean_data**2 / std_data**2
    scale_guess = std_data**2 / mean_data
    params_guess = [shape_guess, scale_guess]
    
    # Optimize the negative log-likelihood
    result = minimize(negative_log_likelihood, params_guess, args=(data))
    
    # Return the estimated parameters
    shape_mle, scale_mle = result.x
    return shape_mle, scale_mle

# Apply the MLE estimation function to the "LL" and "UL" columns
filtered_data["Shape"], filtered_data["Scale"] = zip(*filtered_data.apply(lambda row: estimate_gamma_mle([row["LL"], row["UL"]]), axis=1))

# Export the updated table to a new CSV file
output_path = "/mnt/data/Updated_Microcosting_MLE.csv"
filtered_data[["Test", "Category", "Stage", "LL", "MP", "UL", "Shape", "Scale"]].to_csv(output_path, index=False)
```

```python
import pandas as pd
import numpy as np

# 1. Load the file and specified columns
data = pd.read_csv("/mnt/data/Final_Updated_Microcosting_MLE.csv", usecols=["Test", "Category", "Stage", "LL", "MP", "UL", "Shape", "Scale"])

# 2. For each row, generate a gamma distribution and sample from it
# Initialize lists to store the sampled percentiles
p2_5 = []
p50 = []
p97_5 = []

for index, row in data.iterrows():
    shape, scale = row["Shape"], row["Scale"]
    
    # Check for negative or non-finite values
    if shape > 0 and scale > 0 and np.isfinite(shape) and np.isfinite(scale):
        sample = np.random.gamma(shape, scale, 10000)
        p2_5.append(np.percentile(sample, 2.5))
        p50.append(np.percentile(sample, 50))
        p97_5.append(np.percentile(sample, 97.5))
    else:
        # If the shape or scale is not valid, append NaN
        p2_5.append(np.nan)
        p50.append(np.nan)
        p97_5.append(np.nan)

# 3. Add the sampled percentiles to the dataframe
data["2.5th_percentile"] = p2_5
data["50th_percentile"] = p50
data["97.5th_percentile"] = p97_5

# 4. Export the updated data to a file
output_file_path = "/mnt/data/Updated_Microcosting_With_Percentiles.csv"
data.to_csv(output_file_path, index=False)
```

### Fit Bayesian model, provide 2.5/97.5 and mean

```python
import pandas as pd
import numpy as np
import pymc3 as pm

# Load the data
data = pd.read_csv("/Users/doughnut/Downloads/Final_Updated_Microcosting_MLE.csv", usecols=["Test", "Category", "Stage", "LL", "MP", "UL", "Shape", "Scale"])

# Lists to store the summaries for the posterior distributions
means = []
lower_bound = []
upper_bound = []

for index, row in data.iterrows():
    ll, mp, ul = row['LL'], row['MP'], row['UL']

    with pm.Model() as model:
        # Define the prior using the triangular distribution
        parameter = pm.Triangular('parameter', lower=ll, c=mp, upper=ul)

        # Here we assume a normal likelihood. You need to adjust based on your actual data and problem.
        # If you have an observation for each row, you would put that in 'observed'
        obs = 0  # Placeholder, replace with your actual observation
        likelihood = pm.Normal('likelihood', mu=parameter, sigma=1, observed=obs)

        # Run the MCMC to obtain samples from the posterior
        trace = pm.sample(5000, tune=2000, chains=2)

        # Extract summaries
        means.append(np.mean(trace['parameter']))
        lower_bound.append(np.percentile(trace['parameter'], 2.5))
        upper_bound.append(np.percentile(trace['parameter'], 97.5))

# Store summaries in the dataframe
data["Posterior_Mean"] = means
data["2.5th_percentile"] = lower_bound
data["97.5th_percentile"] = upper_bound

# Export the updated data
output_file_path = "/Users/doughnut/Downloads/Updated_Microcosting_With_Bayesian.csv"
data.to_csv(output_file_path, index=False)
```

```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

# 1. Load the data
data = pd.read_csv('/Users/doughnut/Downloads/Final_Updated_Microcosting_MLE.csv')
selected_data = data[["Test", "Category", "Stage", "LL", "MP", "UL", "Shape", "Scale"]]

# 2. Filter the data
filtered_data = selected_data[(selected_data['Category'] == 'Total') & (selected_data['Stage'] != 'Total')]

# 3. Probabilistic Sensitivity Analysis
samples_sum = {key: [] for key in filtered_data['Test'].unique()}
num_iterations = 10000

for i in range(num_iterations):
    for _, row in filtered_data.iterrows():
        key = row['Test']
        sample_value = np.random.gamma(row['Shape'], row['Scale'])
        samples_sum[key].append(sample_value)

for key in samples_sum.keys():
    samples_sum[key] = np.array(samples_sum[key]).reshape(-1, filtered_data['Test'].value_counts()[key]).sum(axis=1)

# 4. Plotting the results
for key, values in samples_sum.items():
    plt.figure(figsize=(10, 5))
    sns.histplot(values, bins=50, kde=True, color='skyblue')
    plt.title(f"{key} (Probability Distribution)")
    plt.xlabel("Cost (AUD)")
    plt.ylabel("Frequency")
    plt.tight_layout()
    plt.show()
```

```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

# 1. Load the data
data = pd.read_csv('/Users/doughnut/Downloads/Final_Updated_Microcosting_MLE.csv')
selected_data = data[["Test", "Category", "Stage", "LL", "MP", "UL", "Shape", "Scale"]]

# 2. Filter the data
filtered_data = selected_data[(selected_data['Category'] == 'Total') & (selected_data['Stage'] != 'Total')]

# 3. Probabilistic Sensitivity Analysis
samples_sum = {key: [] for key in filtered_data['Test'].unique()}
num_iterations = 10000

for i in range(num_iterations):
    for _, row in filtered_data.iterrows():
        key = row['Test']
        sample_value = np.random.gamma(row['Shape'], row['Scale'])
        samples_sum[key].append(sample_value)

for key in samples_sum.keys():
    samples_sum[key] = np.array(samples_sum[key]).reshape(-1, filtered_data['Test'].value_counts()[key]).sum(axis=1)

# 4. Function to annotate the percentiles on the plot
def annotate_percentiles_final_spacing(values, ax):
    percentiles = [2.5, 50, 97.5]
    for percentile in percentiles:
        value = np.percentile(values, percentile)
        ax.axvline(value, color='red', linestyle='--', alpha=0.7)
        ax.text(value, ax.get_ylim()[1], f'{percentile}%; ', color='red', va='top', ha='right')
        ax.text(value + (0.01 * max(values)), ax.get_ylim()[1], f'{int(value)}', color='red', va='top', ha='left')

# 5. Plotting the results with annotations
for key, values in samples_sum.items():
    plt.figure(figsize=(10, 5))
    ax = sns.histplot(values, bins=50, kde=True, color='skyblue')
    plt.title(f"{key} Cost per Test Probability Distribution")
    plt.xlabel("Cost (AUD)")
    plt.ylabel("Count of Samples")
    annotate_percentiles_final_spacing(values, ax)
    plt.xticks(rotation=30)
    plt.tight_layout()
    plt.show()
```
