```r
# Required Libraries
library(tidyverse)
library(ggplot2)

# Define the instrument cost
X <- 343507.304

# Define utilisation amounts
utilisation <- c("WGS 30x" = 2920, "WES 50x" = 44165, "WES 80x" = 27740)

# Define the utilisation cost for each scenario
scenarios <- list("WEA-T" = c("WES 80x" = 1, "WES 50x" = 2),
                  "WGA-T" = c("WGS 30x" = 3),
                  "Panel" = c("WES 80x" = 1),
                  "WEA-P" = c("WES 80x" = 1),
                  "WGA-P" = c("WGS 30x" = 1))

# Create DataFrame
df <- data.frame()

for (scenario in names(scenarios)) {
  for (percent in seq(30, 100, 10)) {
    cost <- 0
    for (util in names(scenarios[[scenario]])) {
      n <- scenarios[[scenario]][util]
      cost <- cost + X / ((utilisation[util] * n) * percent / 100)
    }
    df <- rbind(df, data.frame(Scenario = scenario, Utilisation_Percent = percent, Cost = cost))
  }
}

# Plotting
ggplot(df, aes(x = Utilisation_Percent, y = Cost, color = Scenario)) +
  geom_line() +
  labs(title = 'Cost vs Utilisation % for different Scenarios', x = 'Utilisation %') +
  theme_minimal()
```
