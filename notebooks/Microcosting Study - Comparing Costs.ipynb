```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np

# Data setup
components = [
    "Specimen Collection",
    "Specimen Reception",
    "DNA Extraction",
    "Library Preparation",
    "Sequencing",
    "Processing",
    "Storage",
    "Filtering",
    "Interpretation",
    "Report Writing",
]
groups = ["Panel", "EA-P", "EA-T", "WGA-P", "WGA-T"]
values = [
    [26, 71, 58, 411, 1042, 22, 3, 566, 32, 140],
    [26, 74, 58, 411, 1042, 22, 1, 757, 291, 140],
    [73, 71, 174, 1234, 3046, 66, 8, 566, 291, 140],
    [26, 74, 58, 315, 2683, 121, 336, 796, 291, 140],
    [73, 74, 152, 837, 8048, 362, 1007, 605, 291, 140],
]
df = pd.DataFrame(values, columns=components, index=groups)
df = df.reset_index().melt(id_vars="index", var_name="component", value_name="value")

# Filtering
df_wga = df[df["index"].isin(["WGA-P", "WGA-T"])]
df_rest = df[~df["index"].isin(["WGA-P", "WGA-T"])]

# Custom palette
custom_palette = {
    "Panel": "#1f77b4",
    "EA-P": "#ff7f0e",
    "EA-T": "#2ca02c",
    "WGA-P": "#d62728",
    "WGA-T": "#9467bd",
}

# Define the number of components and groups
num_components = len(components)
num_groups = len(groups)

# Define the width for each bar and positions for each component group
bar_width = 0.15
positions = np.arange(num_components)

# Set up the figure and axes
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(13, 7))

# Create secondary y-axes for values beyond the specified thresholds
ax1_secondary = ax1.twinx()
ax2_secondary = ax2.twinx()


# Function to plot bars for each group on a given axis with threshold-based switching to secondary axis
def plot_bars_with_threshold(df, ax, ax_secondary, threshold, title):
    handles, labels = [], []  # Collect handles and labels for legend
    for idx, group in enumerate(df["index"].unique()):
        group_data = df[df["index"] == group]
        for pos, value in zip(positions, group_data["value"]):
            if value <= threshold:
                bar = ax.bar(
                    pos + idx * bar_width,
                    value,
                    bar_width,
                    label=group,
                    color=custom_palette[group],
                )
            else:
                bar = ax_secondary.bar(
                    pos + idx * bar_width, value, bar_width, color=custom_palette[group]
                )

            # Add to handles and labels if it's a new group
            if group not in labels:
                handles.append(bar[0])
                labels.append(group)

    ax.set_xticks(positions + bar_width * (num_groups / 2))
    ax.set_xticklabels(components, rotation=30, ha="right")
    ax.set_ylabel("Cost (AUD $)")
    ax.set_xlabel("Stage")
    ax.grid(which="major", alpha=0.5)
    ax_secondary.grid(None)  # Turn off grid for the secondary y-axis
    return handles, labels


# Plotting with threshold adjustments and gathering handles and labels for the legend
handles1, labels1 = plot_bars_with_threshold(
    df_wga, ax1, ax1_secondary, 2000, "WGA-T and WGA-P"
)
handles2, labels2 = plot_bars_with_threshold(
    df_rest, ax2, ax2_secondary, 1000, "Other Groups"
)

# Combining handles and labels from both subplots
all_handles = handles1 + handles2
all_labels = labels1 + labels2

# Adding a unified legend to the figure at the bottom
fig.legend(
    all_handles,
    all_labels,
    loc="lower center",
    bbox_to_anchor=(0.5, -0.05),
    ncol=5,
    title="Genomic Test",
    frameon=False,
)

# Adjusting layout
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np

# Data setup
components = [
    "Specimen Collection",
    "Specimen Reception",
    "DNA Extraction",
    "Library Preparation",
    "Sequencing",
    "Processing",
    "Storage",
    "Filtering",
    "Interpretation",
    "Report Writing",
]
groups = ["Panel", "EA-P", "EA-T", "WGA-P", "WGA-T"]
values = [
    [26, 71, 58, 411, 1042, 22, 3, 566, 32, 140],
    [26, 74, 58, 411, 1042, 22, 1, 757, 291, 140],
    [73, 71, 174, 1234, 3046, 66, 8, 566, 291, 140],
    [26, 74, 58, 315, 2683, 121, 336, 796, 291, 140],
    [73, 74, 152, 837, 8048, 362, 1007, 605, 291, 140],
]
df = pd.DataFrame(values, columns=components, index=groups)
df = df.reset_index().melt(id_vars="index", var_name="component", value_name="value")

# Filtering
df_wga = df[df["index"].isin(["WGA-P", "WGA-T"])]
df_rest = df[~df["index"].isin(["WGA-P", "WGA-T"])]

# Custom palette
custom_palette = {
    "Panel": "#1f77b4",
    "EA-P": "#ff7f0e",
    "EA-T": "#2ca02c",
    "WGA-P": "#d62728",
    "WGA-T": "#9467bd",
}

# Matplotlib plot, fit to A4 size (8.27x11.69 inches) considering some margins
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(8.27, 11.69 - 2))

# Define the number of components and groups
num_components = len(components)
num_groups = len(groups)

# Define the width for each bar and positions for each component group
bar_width = 0.15
positions = np.arange(num_components)


# Function to plot bars for each group on a given axis
def plot_bars(df, ax, title):
    for idx, group in enumerate(df["index"].unique()):
        group_data = df[df["index"] == group]
        ax.bar(
            positions + idx * bar_width,
            group_data["value"],
            bar_width,
            label=group,
            color=custom_palette[group],
        )
    ax.set_xticks(positions + bar_width * (num_groups / 2))
    ax.set_xticklabels(components, rotation=45, ha="right")  # Label at 45 degrees
    ax.set_ylabel("Cost (AUD $)")
    ax.set_xlabel("Stage")
    ax.legend(title="Genomic Test", loc="upper right")
    ax.grid(which="major", alpha=0.5)
    ax.yaxis.set_major_locator(mticker.MultipleLocator(500))  # Major y-ticks every 500
    ax.yaxis.set_minor_locator(mticker.MultipleLocator(100))  # Minor y-ticks every 100
    ax.grid(which="minor", alpha=0.2)


plot_bars(df_wga, ax1, "WGA-T and WGA-P")
plot_bars(df_rest, ax2, "Other Groups")

# Set major ticks for right subplot (ax2) to 250
ax2.yaxis.set_major_locator(
    mticker.MultipleLocator(250)
)  # Major y-ticks every 250 for ax2
ax2.yaxis.set_minor_locator(
    mticker.MultipleLocator(50)
)  # Minor y-ticks every 50 for ax2


# Other plot adjustments, legends, tick marks, etc.
plt.tight_layout()
plt.show()
```
